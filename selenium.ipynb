```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))

mobiles_data = []

total_pages = 3

for page in range(1, total_pages + 1):
    url = f"https://www.noon.com/egypt-ar/electronics-and-mobiles/mobiles-and-accessories/mobiles-20905/?limit=50&page={page}&sort[by]=popularity&sort[dir]=desc"
    driver.get(url)
    time.sleep(4)

    cards = driver.find_elements(By.CLASS_NAME, "_productBoxLink_1ts6x_5")

    for card in cards:
        try:
            price = card.find_element(By.CLASS_NAME, "_amount_layq2_58").text
        except:
            price = ""

        try:
            old_price = card.find_element(By.CLASS_NAME, "_oldPrice_layq2_84").text
        except:
            old_price = ""

        try:
            discount = card.find_element(By.CLASS_NAME, "_discount_grckm_1").text
        except:
            discount = ""

        try:
            rating_count = card.find_element(By.CLASS_NAME, "_countCtr_1r83y_47").text
        except:
            rating_count = ""

        try:
            img_elements = card.find_elements(By.CLASS_NAME, "_productImage_1yztz_82")
            images_list = [img.get_attribute("src") for img in img_elements if img.get_attribute("src")]
            all_images = ", ".join(images_list)
        except:
            all_images = ""

        try:
            nudge_info = card.find_element(By.CLASS_NAME, "_nudgesContainer_1f9au_1").text
        except:
            nudge_info = ""

        mobiles_data.append({
            "Price": price,
            "Old_Price": old_price,
            "Discount": discount,
            "Rating_Count": rating_count,
            "Images": all_images,
            "Status_And_Offers": nudge_info
        })



df = pd.DataFrame(mobiles_data)
df.to_csv("noon_data.csv", index=False, encoding="utf-8-sig")
df
```
